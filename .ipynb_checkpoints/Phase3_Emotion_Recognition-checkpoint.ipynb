# SECTION 3 — Emotion Labels Module

`emotion_labels.py` defines the canonical 7-class emotion set this module supports —
**Happy, Sad, Neutral, Angry, Fear, Surprise, Disgust** — matching the class names
DeepFace's `Emotion` model produces, so no remapping is needed between the raw model
output and what the rest of the pipeline consumes. It also provides a
teacher-facing display label (with an icon) and a coarse positive/neutral/negative
polarity for each emotion, which Phase 5's engagement-score engine will use.

Requirements covered: the 7 emotion classes requested in the brief, a validation
helper, and reusable display metadata.

````python
"""
Section 3 -- Write models/emotion_detection/emotion_labels.py
"""
emotion_labels_code = '''"""
Canonical emotion label definitions for the Emotion Recognition subsystem.

This is the single source of truth for the 7 emotion classes supported
throughout Phase 3 and consumed by later phases (engagement scoring,
cognitive state classification). Every other Phase 3 module imports
its label constants from here rather than hard-coding strings.
"""
from dataclasses import dataclass
from typing import Dict, List

from utils.logger import get_logger

logger = get_logger(__name__)


# Order matches the class order DeepFace's "Emotion" model reports scores in.
EMOTION_LABELS: List[str] = [
    "angry",
    "disgust",
    "fear",
    "happy",
    "sad",
    "surprise",
    "neutral",
]


@dataclass(frozen=True)
class EmotionMeta:
    """Display metadata for one emotion label."""
    label: str
    display_name: str
    icon: str
    polarity: str  # "positive" | "neutral" | "negative"


EMOTION_METADATA: Dict[str, EmotionMeta] = {
    "angry":    EmotionMeta("angry",    "Angry",    "\U0001F620", "negative"),
    "disgust":  EmotionMeta("disgust",  "Disgust",  "\U0001F922", "negative"),
    "fear":     EmotionMeta("fear",     "Fear",     "\U0001F628", "negative"),
    "happy":    EmotionMeta("happy",    "Happy",    "\U0001F642", "positive"),
    "sad":      EmotionMeta("sad",      "Sad",      "\U0001F622", "negative"),
    "surprise": EmotionMeta("surprise", "Surprise", "\U0001F62E", "neutral"),
    "neutral":  EmotionMeta("neutral",  "Neutral",  "\U0001F610", "neutral"),
}


class InvalidEmotionLabelError(ValueError):
    """Raised when a label outside EMOTION_LABELS is used where a valid one is required."""


def is_valid_label(label: str) -> bool:
    """Check whether a string is one of the 7 supported emotion labels.

    Args:
        label: Candidate label, matched case-insensitively.

    Returns:
        True if the label is supported, False otherwise.
    """
    return label.lower() in EMOTION_LABELS


def validate_label(label: str) -> str:
    """Validate and normalize an emotion label to lowercase.

    Args:
        label: Candidate label.

    Returns:
        The normalized (lowercase) label.

    Raises:
        InvalidEmotionLabelError: If the label is not one of EMOTION_LABELS.
    """
    normalized = label.lower().strip()
    if normalized not in EMOTION_LABELS:
        raise InvalidEmotionLabelError(
            f"'{label}' is not a supported emotion label. Expected one of {EMOTION_LABELS}."
        )
    return normalized


def get_display_name(label: str) -> str:
    """Return the teacher-facing display name for an emotion label."""
    normalized = validate_label(label)
    return EMOTION_METADATA[normalized].display_name


def get_polarity(label: str) -> str:
    """Return the coarse polarity ("positive"/"neutral"/"negative") for an emotion label."""
    normalized = validate_label(label)
    return EMOTION_METADATA[normalized].polarity
'''

emotion_labels_path = emotion_dir / "emotion_labels.py"
emotion_labels_path.write_text(
    emotion_labels_code,
    encoding="utf-8"
)
print(f"Written: {emotion_labels_path.relative_to(project_root)}")
````

# SECTION 4 — Emotion Preprocessing Module

`emotion_preprocessing.py` prepares a face crop for emotion analysis. Rather than
reimplementing blur/brightness/size quality gating, it **reuses Phase 2's
`preprocessing.check_quality()` and `preprocessing.pad_to_square()`** directly — the
same face crop that already passed Phase 2's gate before embedding is a good candidate
for emotion analysis too, so the gate is shared rather than duplicated.

On top of that shared gate, this module adds emotion-model-specific preparation: resize
to the configured input size, optional grayscale conversion (the classic FER-2013 input
format), and float32 normalization to `[0, 1]`.

Requirements covered: resize, normalization, reuse of Phase 2 quality gating, and a
dedicated `EmotionPreprocessingError` for bad input.

````python
"""
Section 4 -- Write models/emotion_detection/emotion_preprocessing.py
"""
emotion_preprocessing_code = '''"""
Face-crop preprocessing for the Emotion Recognition subsystem.

Quality gating (blur/brightness/size) is deliberately NOT reimplemented
here -- it is reused directly from Phase 2's
models.face_authentication.preprocessing module, so a face crop is
held to the exact same quality bar before either embedding (Phase 2)
or emotion analysis (Phase 3).
"""
from typing import Tuple

import cv2
import numpy as np

from models.face_authentication.preprocessing import check_quality, pad_to_square, QualityReport
from utils.logger import get_logger

logger = get_logger(__name__)


class EmotionPreprocessingError(Exception):
    """Raised when a face crop cannot be prepared for emotion analysis."""


def preprocess_for_emotion(
    face_bgr: np.ndarray,
    target_size: Tuple[int, int] = (48, 48),
    grayscale: bool = True,
    enforce_quality: bool = True,
) -> np.ndarray:
    """Prepare a raw BGR face crop for emotion classification.

    Args:
        face_bgr: Cropped face region straight from face detection, BGR uint8.
        target_size: (width, height) the emotion model expects. Defaults to
            48x48, the classic FER-2013 input size.
        grayscale: If True (default), convert to single-channel grayscale,
            matching the FER-2013 training format. If False, RGB is kept
            (DeepFace's own analyze() call handles color internally, so this
            flag mainly matters for custom/offline models).
        enforce_quality: If True (default), reject crops that fail Phase 2's
            check_quality() gate (too blurry, too dark/bright, too small).

    Returns:
        A float32 array normalized to [0, 1], shape (H, W) if grayscale else
        (H, W, 3).

    Raises:
        EmotionPreprocessingError: If the crop is empty, fails the quality
            gate (when enforced), or cannot be resized.
    """
    if face_bgr is None or face_bgr.size == 0:
        raise EmotionPreprocessingError("Received an empty face crop.")

    if enforce_quality:
        report: QualityReport = check_quality(face_bgr)
        if not report.passed:
            raise EmotionPreprocessingError(
                f"Face crop failed quality gate for emotion analysis: {report.reasons}"
            )

    squared = pad_to_square(face_bgr)

    try:
        resized = cv2.resize(squared, target_size, interpolation=cv2.INTER_AREA)
    except cv2.error as exc:
        raise EmotionPreprocessingError(f"Failed to resize face crop: {exc}") from exc

    if grayscale:
        gray = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)
        normalized = gray.astype(np.float32) / 255.0
    else:
        rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)
        normalized = rgb.astype(np.float32) / 255.0

    return normalized


def to_model_input(normalized: np.ndarray) -> np.ndarray:
    """Add the batch (and channel, if grayscale) dimensions a CNN expects.

    Args:
        normalized: Output of preprocess_for_emotion(), shape (H, W) or (H, W, 3).

    Returns:
        A 4-D array of shape (1, H, W, 1) or (1, H, W, 3), ready for a
        batch-of-one CNN forward pass.
    """
    if normalized.ndim == 2:
        normalized = normalized[..., np.newaxis]
    return normalized[np.newaxis, ...]
'''

emotion_preprocessing_path = emotion_dir / "emotion_preprocessing.py"
emotion_preprocessing_path.write_text(emotion_preprocessing_code)
print(f"Written: {emotion_preprocessing_path.relative_to(project_root)}")
````

````python
"""
Section 4 -- Test the emotion preprocessing module against a good
crop (should pass end-to-end) and a too-small/too-dark crop (should
raise EmotionPreprocessingError via the reused Phase 2 quality gate).
"""
import importlib
import numpy as np
import models.emotion_detection.emotion_preprocessing as emotion_preprocessing_module
importlib.reload(emotion_preprocessing_module)
from models.emotion_detection.emotion_preprocessing import (
    preprocess_for_emotion, to_model_input, EmotionPreprocessingError,
)

rng = np.random.default_rng(0)
good_crop = rng.integers(90, 200, size=(160, 140, 3), dtype=np.uint8)

processed_gray = preprocess_for_emotion(good_crop, target_size=(48, 48), grayscale=True)
assert processed_gray.shape == (48, 48)
assert processed_gray.dtype == np.float32
assert 0.0 <= processed_gray.min() and processed_gray.max() <= 1.0
print(f"PASS: grayscale preprocessing -- shape={processed_gray.shape}, "
      f"range=[{processed_gray.min():.3f}, {processed_gray.max():.3f}]")

processed_rgb = preprocess_for_emotion(good_crop, target_size=(224, 224), grayscale=False)
assert processed_rgb.shape == (224, 224, 3)
print(f"PASS: RGB preprocessing -- shape={processed_rgb.shape}")

model_input = to_model_input(processed_gray)
assert model_input.shape == (1, 48, 48, 1)
print(f"PASS: to_model_input() adds batch/channel dims -- shape={model_input.shape}")

dark_crop = np.full((150, 130, 3), 5, dtype=np.uint8)
try:
    preprocess_for_emotion(dark_crop, enforce_quality=True)
    raise AssertionError("Expected EmotionPreprocessingError was not raised.")
except EmotionPreprocessingError as exc:
    print(f"PASS: EmotionPreprocessingError correctly raised for a too-dark crop -- {exc}")

try:
    preprocess_for_emotion(np.zeros((0, 0, 3), dtype=np.uint8))
    raise AssertionError("Expected EmotionPreprocessingError was not raised for empty crop.")
except EmotionPreprocessingError as exc:
    print(f"PASS: EmotionPreprocessingError correctly raised for an empty crop -- {exc}")
````

# SECTION 5 — Emotion Utils Module

`emotion_utils.py` provides small, pure-function helpers shared by the detector, the
router, and later phases:

- `normalize_scores()` — turns DeepFace's raw per-class score dict (which does not
  always sum to exactly 1.0) into a proper probability distribution.
- `top_emotion()` — picks the single highest-confidence label from a score dict.
- `EmotionSmoother` — an exponential-moving-average smoother that reduces frame-to-frame
  jitter in live video, so the emotion shown/stored doesn't flicker between "neutral"
  and "happy" every other frame.
- `summarize_session()` — aggregates a list of per-frame results into a
  session-level emotion distribution, used by the `/emotion/predict` router's callers
  and by later reporting phases.

````python
"""
Section 5 -- Write models/emotion_detection/emotion_utils.py
"""
emotion_utils_code = '''"""
Shared helper functions for the Emotion Recognition subsystem: score
normalization, temporal smoothing, and session-level aggregation.
"""
from collections import Counter, deque
from dataclasses import dataclass
from typing import Deque, Dict, List, Optional, Tuple

from models.emotion_detection.emotion_labels import EMOTION_LABELS, validate_label
from utils.logger import get_logger

logger = get_logger(__name__)


def normalize_scores(raw_scores: Dict[str, float]) -> Dict[str, float]:
    """Normalize a raw per-class score dict into a probability distribution.

    Args:
        raw_scores: Mapping of emotion label -> raw score (e.g. percentages
            from DeepFace, which do not always sum to exactly 100).

    Returns:
        A new dict with the same keys, values rescaled to sum to 1.0.

    Raises:
        ValueError: If raw_scores is empty or every value is zero.
    """
    if not raw_scores:
        raise ValueError("raw_scores must not be empty.")
    total = sum(raw_scores.values())
    if total <= 0:
        raise ValueError("raw_scores must contain at least one positive value.")
    return {label: value / total for label, value in raw_scores.items()}


def top_emotion(scores: Dict[str, float]) -> Tuple[str, float]:
    """Return the (label, confidence) pair with the highest score.

    Args:
        scores: Mapping of emotion label -> score (normalized or raw).

    Returns:
        A (label, confidence) tuple for the highest-scoring emotion.

    Raises:
        ValueError: If scores is empty.
    """
    if not scores:
        raise ValueError("scores must not be empty.")
    label = max(scores, key=scores.get)
    return label, scores[label]


class EmotionSmoother:
    """Exponential-moving-average smoother for per-frame emotion scores.

    Reduces frame-to-frame flicker in live video by blending each new
    reading with the running average rather than trusting a single frame.

    Example:
        >>> smoother = EmotionSmoother(alpha=0.3)
        >>> smoothed = smoother.update({"happy": 0.9, "neutral": 0.1})
    """

    def __init__(self, alpha: float = 0.3) -> None:
        """
        Args:
            alpha: Weight given to the newest reading, in (0, 1]. Higher
                values track new readings faster; lower values smooth more.
        """
        if not (0.0 < alpha <= 1.0):
            raise ValueError("alpha must be in (0, 1].")
        self.alpha = alpha
        self._state: Optional[Dict[str, float]] = None

    def update(self, scores: Dict[str, float]) -> Dict[str, float]:
        """Blend a new frame's scores into the running smoothed estimate.

        Args:
            scores: This frame's (typically normalized) per-class scores.

        Returns:
            The updated smoothed score dict.
        """
        normalized = normalize_scores(scores)
        if self._state is None:
            self._state = dict(normalized)
        else:
            self._state = {
                label: self.alpha * normalized.get(label, 0.0)
                       + (1 - self.alpha) * self._state.get(label, 0.0)
                for label in EMOTION_LABELS
            }
        return dict(self._state)

    def reset(self) -> None:
        """Clear the smoother's running state (e.g. on a new session/student)."""
        self._state = None


@dataclass
class SessionEmotionSummary:
    """Aggregated emotion statistics over a sequence of frames."""
    dominant_emotion: str
    emotion_counts: Dict[str, int]
    emotion_percentages: Dict[str, float]
    num_frames: int


def summarize_session(frame_labels: List[str]) -> SessionEmotionSummary:
    """Aggregate a sequence of per-frame top-emotion labels into a summary.

    Args:
        frame_labels: List of emotion labels, one per analyzed frame
            (e.g. collected over a class session).

    Returns:
        A SessionEmotionSummary describing the dominant emotion and the
        full distribution.

    Raises:
        ValueError: If frame_labels is empty, or contains an unsupported label.
    """
    if not frame_labels:
        raise ValueError("frame_labels must not be empty.")
    validated = [validate_label(label) for label in frame_labels]
    counts = Counter(validated)
    total = len(validated)
    percentages = {label: (count / total) * 100.0 for label, count in counts.items()}
    dominant = counts.most_common(1)[0][0]
    return SessionEmotionSummary(
        dominant_emotion=dominant,
        emotion_counts=dict(counts),
        emotion_percentages=percentages,
        num_frames=total,
    )
'''

emotion_utils_path = emotion_dir / "emotion_utils.py"
emotion_utils_path.write_text(emotion_utils_code)
print(f"Written: {emotion_utils_path.relative_to(project_root)}")
````

````python
"""
Section 5 -- Test the emotion utils module: score normalization, top
emotion selection, the EmotionSmoother's convergence behaviour, and
session-level aggregation.
"""
import importlib
import models.emotion_detection.emotion_utils as emotion_utils_module
importlib.reload(emotion_utils_module)
from models.emotion_detection.emotion_utils import (
    normalize_scores, top_emotion, EmotionSmoother, summarize_session,
)

raw = {"happy": 72.0, "neutral": 20.0, "sad": 8.0}
normalized = normalize_scores(raw)
assert abs(sum(normalized.values()) - 1.0) < 1e-9
print(f"PASS: normalize_scores -- {normalized}")

label, confidence = top_emotion(normalized)
assert label == "happy"
print(f"PASS: top_emotion -- label={label}, confidence={confidence:.3f}")

smoother = EmotionSmoother(alpha=0.5)
first = smoother.update({"happy": 1.0})
second = smoother.update({"happy": 1.0})
assert second["happy"] >= first["happy"], "Smoother should remain stable or converge toward repeated readings"
smoother.reset()
after_reset = smoother.update({"neutral": 1.0})
assert after_reset["neutral"] == 1.0, "Smoother should start fresh after reset()"
print(f"PASS: EmotionSmoother converges ({first['happy']:.3f} -> {second['happy']:.3f}) and resets cleanly")

summary = summarize_session(["happy", "happy", "neutral", "sad", "happy"])
assert summary.dominant_emotion == "happy"
assert summary.num_frames == 5
assert summary.emotion_counts["happy"] == 3
print(f"PASS: summarize_session -- dominant={summary.dominant_emotion}, "
      f"counts={summary.emotion_counts}, percentages={ {k: round(v,1) for k,v in summary.emotion_percentages.items()} }")
````

# SECTION 6 — Emotion Detector Module

`emotion_detector.py` wraps **DeepFace's emotion analysis backbone** (a
MobileNetV2/CNN-based FER model bundled with DeepFace) behind an `EmotionDetector`
class. DeepFace and TensorFlow are imported **lazily inside `__init__`**, mirroring
Phase 2's `FaceEmbedder` exactly: modules that never construct an `EmotionDetector`
(e.g. `emotion_labels.py`, `emotion_utils.py`) never pay the TensorFlow import cost, and
if DeepFace/TensorFlow is unavailable in the current environment the failure is raised
as a typed `EmotionAnalysisError` at construction time — never an uncaught crash, and
never at import time.

Supports: Happy, Sad, Neutral, Angry, Fear, Surprise, Disgust — the full 7-class set
from `emotion_labels.py`. Every call also reports `processing_time` in milliseconds, so
the router can surface it directly.

````python
"""
Section 6 -- Write models/emotion_detection/emotion_detector.py
"""
emotion_detector_code = '''"""
Emotion detection module built on DeepFace's bundled emotion (FER) backbone.
"""
import time
from dataclasses import dataclass
from typing import Dict, Optional

import numpy as np

from models.emotion_detection.emotion_labels import EMOTION_LABELS
from models.emotion_detection.emotion_utils import normalize_scores, top_emotion
from utils.logger import get_logger

logger = get_logger(__name__)


class EmotionAnalysisError(Exception):
    """Raised when an emotion prediction could not be generated for a face image."""


@dataclass
class EmotionResult:
    """Result of running emotion analysis on a single face image."""
    emotion: str
    confidence: float
    scores: Dict[str, float]
    processing_time_ms: float


class EmotionDetector:
    """Classifies a face crop into one of the 7 supported emotions.

    DeepFace is imported lazily inside __init__ so that modules which do
    not need emotion analysis never pay the TensorFlow/DeepFace import
    cost -- the same defensive pattern Phase 2's FaceEmbedder used for
    the Facenet512 embedding backbone.
    """

    def __init__(self, confidence_threshold: float = 0.4) -> None:
        """
        Args:
            confidence_threshold: Minimum top-emotion confidence (0-1) for
                a prediction to be considered reliable. Predictions below
                this threshold are still returned, but flagged via
                EmotionResult.confidence for the caller to act on.

        Raises:
            EmotionAnalysisError: If DeepFace/TensorFlow is not installed.
        """
        try:
            from deepface import DeepFace  # noqa: WPS433 -- deliberate lazy import
        except ImportError as exc:
            raise EmotionAnalysisError(
                "DeepFace is not installed. Run `pip install deepface tensorflow` "
                "(see Phase 1 Section 4 special-installation notes)."
            ) from exc

        self._DeepFace = DeepFace
        self.confidence_threshold = confidence_threshold
        # Trigger a one-time model build/download so later calls are fast.
        try:
            self._DeepFace.build_model("Emotion")
        except Exception as exc:  # noqa: BLE001 -- DeepFace raises assorted exception types
            raise EmotionAnalysisError(f"Failed to load the DeepFace Emotion model: {exc}") from exc
        logger.info("EmotionDetector ready -- backbone=DeepFace Emotion (7-class FER)")

    def analyze(self, face_bgr: np.ndarray) -> EmotionResult:
        """Predict the emotion of a single face image.

        Args:
            face_bgr: Cropped face region, BGR uint8, as produced by
                Phase 2's face detection (detection/quality gating is the
                caller's responsibility -- this method trusts its input).

        Returns:
            An EmotionResult with the top emotion, its confidence, the
            full 7-class score distribution, and processing time in ms.

        Raises:
            EmotionAnalysisError: If DeepFace fails to produce a prediction.
        """
        if face_bgr is None or face_bgr.size == 0:
            raise EmotionAnalysisError("Received an empty face image.")

        start = time.perf_counter()
        try:
            result = self._DeepFace.analyze(
                img_path=face_bgr,
                actions=["emotion"],
                enforce_detection=False,  # we already ran our own MediaPipe detector
                detector_backend="skip",
            )
        except Exception as exc:  # noqa: BLE001 -- DeepFace raises assorted exception types
            raise EmotionAnalysisError(f"DeepFace failed to analyze emotion: {exc}") from exc
        elapsed_ms = (time.perf_counter() - start) * 1000.0

        payload = result[0] if isinstance(result, list) else result
        raw_scores = payload.get("emotion")
        if not raw_scores:
            raise EmotionAnalysisError("DeepFace returned no emotion scores for the given face.")

        scores = normalize_scores({k.lower(): float(v) for k, v in raw_scores.items()})
        missing = set(EMOTION_LABELS) - set(scores)
        for label in missing:
            scores[label] = 0.0

        label, confidence = top_emotion(scores)
        if confidence < self.confidence_threshold:
            logger.info(
                f"Emotion prediction below confidence threshold: {label}={confidence:.3f} "
                f"< {self.confidence_threshold}"
            )

        return EmotionResult(
            emotion=label, confidence=confidence, scores=scores, processing_time_ms=elapsed_ms,
        )

    def analyze_safe(self, face_bgr: np.ndarray) -> Optional[EmotionResult]:
        """Same as analyze(), but returns None instead of raising.

        Useful in real-time loops where one bad frame must never stop
        the pipeline -- exactly the "never crash" requirement for this
        module.

        Args:
            face_bgr: Cropped face region, BGR uint8.

        Returns:
            An EmotionResult, or None if analysis failed for any reason.
        """
        try:
            return self.analyze(face_bgr)
        except EmotionAnalysisError as exc:
            logger.warning(f"analyze_safe: emotion analysis failed, returning None -- {exc}")
            return None
'''

emotion_detector_path = emotion_dir / "emotion_detector.py"
emotion_detector_path.write_text(emotion_detector_code)
print(f"Written: {emotion_detector_path.relative_to(project_root)}")
````

# SECTION 8 — FastAPI Integration

We add a new router, `backend/app/routers/emotion.py`, exposing:

| Method | Path | Purpose |
|---|---|---|
| `POST` | `/emotion/predict` | Analyze a base64-encoded face crop and return emotion, confidence, processing_time |
| `GET`  | `/emotion/labels`  | List the 7 supported emotion labels and their display metadata |

`backend/app/main.py` is then **patched, not rewritten** — the same idempotent
find-and-insert approach Phase 2 used for `face_authentication`, guarded so re-running
this notebook never registers the router twice.

````python
"""
Section 8 -- Write backend/app/routers/emotion.py

The router constructs EmotionDetector lazily (module import must never
crash the whole API just because DeepFace/TensorFlow isn't installed
yet), and returns a clear 503 if the model backend is genuinely
unavailable at request time -- it never lets an unhandled exception
propagate out of an endpoint.
"""
emotion_router_code = '''"""
Emotion Recognition router.
"""
import base64
from pathlib import Path
from typing import List, Optional

import cv2
import numpy as np
from fastapi import APIRouter, HTTPException
from pydantic import BaseModel, Field

from models.emotion_detection.emotion_detector import EmotionDetector, EmotionAnalysisError
from models.emotion_detection.emotion_labels import EMOTION_LABELS, EMOTION_METADATA
from utils.logger import get_logger

logger = get_logger(__name__)

router = APIRouter(prefix="/emotion", tags=["emotion-recognition"])

# NOTE: EmotionDetector loads a DeepFace/TensorFlow model, which can be slow
# and requires an optional heavy dependency. We deliberately do NOT construct
# it at import time -- a missing dependency must never crash the whole API on
# startup. Instead it is created lazily, on first actual use, and cached.
_detector: Optional[EmotionDetector] = None


def _get_detector() -> EmotionDetector:
    global _detector
    if _detector is None:
        _detector = EmotionDetector()
    return _detector


class PredictRequest(BaseModel):
    face_crop_b64: str = Field(..., description="Base64-encoded JPEG/PNG face crop.")
    student_id: Optional[str] = Field(None, description="Optional student ID for logging/correlation.")


class PredictResponse(BaseModel):
    emotion: str
    confidence: float
    processing_time: float
    scores: dict


class EmotionLabelInfo(BaseModel):
    label: str
    display_name: str
    icon: str
    polarity: str


def _decode_b64_image(face_crop_b64: str) -> Optional[np.ndarray]:
    """Decode a single base64 JPEG/PNG string into a BGR NumPy array."""
    try:
        raw_bytes = base64.b64decode(face_crop_b64)
        array = np.frombuffer(raw_bytes, dtype=np.uint8)
        return cv2.imdecode(array, cv2.IMREAD_COLOR)
    except Exception as exc:  # noqa: BLE001 -- malformed client input takes many shapes
        logger.warning(f"Failed to decode face_crop_b64: {exc}")
        return None


@router.post("/predict", response_model=PredictResponse)
async def predict_emotion(payload: PredictRequest) -> PredictResponse:
    """Analyze a base64-encoded face crop and return the predicted emotion."""
    image = _decode_b64_image(payload.face_crop_b64)
    if image is None:
        raise HTTPException(status_code=400, detail="face_crop_b64 could not be decoded as an image.")

    try:
        detector = _get_detector()
    except EmotionAnalysisError as exc:
        raise HTTPException(status_code=503, detail=f"Emotion analysis backend unavailable: {exc}") from exc

    try:
        result = detector.analyze(image)
    except EmotionAnalysisError as exc:
        raise HTTPException(status_code=422, detail=f"Emotion analysis failed: {exc}") from exc

    if payload.student_id:
        logger.info(
            f"Emotion prediction for student_id={payload.student_id}: "
            f"{result.emotion} ({result.confidence:.3f})"
        )

    return PredictResponse(
        emotion=result.emotion,
        confidence=result.confidence,
        processing_time=result.processing_time_ms,
        scores=result.scores,
    )


@router.get("/labels", response_model=List[EmotionLabelInfo])
async def list_emotion_labels() -> List[EmotionLabelInfo]:
    """List the 7 supported emotion labels with their display metadata."""
    return [
        EmotionLabelInfo(
            label=label,
            display_name=EMOTION_METADATA[label].display_name,
            icon=EMOTION_METADATA[label].icon,
            polarity=EMOTION_METADATA[label].polarity,
        )
        for label in EMOTION_LABELS
    ]
'''

backend_routers_dir = project_root / "backend" / "app" / "routers"
(backend_routers_dir / "emotion.py").write_text(emotion_router_code)
print(f"Written: {(backend_routers_dir / 'emotion.py').relative_to(project_root)}")
````

````python
"""
Section 8 -- Safely patch backend/app/main.py to register the new
emotion router, WITHOUT rewriting the file from scratch and WITHOUT
duplicating the registration if this notebook is re-run. Mirrors
Phase 2's exact patch strategy for face_authentication, and is
idempotent whether Phase 2's patch, Phase 3's patch, both, or neither
have already been applied.
"""
main_py_path = project_root / "backend" / "app" / "main.py"
main_py_source = main_py_path.read_text()

IMPORT_LINE = "from backend.app.routers import emotion"
INCLUDE_LINE = "    app.include_router(emotion.router)"

already_patched = IMPORT_LINE in main_py_source and INCLUDE_LINE.strip() in main_py_source

if already_patched:
    print("backend/app/main.py already includes the emotion router -- no changes made.")
else:
    patched_source = main_py_source

    # Prefer anchoring after the Phase 2 face_authentication import/include if
    # present (keeps router registration order matching pipeline order:
    # face auth -> emotion), falling back to the Phase 1 health anchor.
    if "from backend.app.routers import face_authentication" in patched_source:
        patched_source = patched_source.replace(
            "from backend.app.routers import face_authentication",
            f"from backend.app.routers import face_authentication\n{IMPORT_LINE}",
        )
        patched_source = patched_source.replace(
            "    app.include_router(face_authentication.router)",
            f"    app.include_router(face_authentication.router)\n{INCLUDE_LINE}",
        )
    elif "from backend.app.routers import health" in patched_source:
        patched_source = patched_source.replace(
            "from backend.app.routers import health",
            f"from backend.app.routers import health\n{IMPORT_LINE}",
        )
        patched_source = patched_source.replace(
            "    app.include_router(health.router)",
            f"    app.include_router(health.router)\n{INCLUDE_LINE}",
        )

    if patched_source == main_py_source:
        raise RuntimeError(
            "Could not locate an expected anchor line in main.py -- "
            "aborting patch to avoid corrupting the file. Check main.py manually."
        )
    main_py_path.write_text(patched_source)
    print("backend/app/main.py patched: emotion router registered.")

print("\n--- relevant lines in main.py ---")
for line in main_py_path.read_text().splitlines():
    if "emotion" in line or "face_authentication" in line or "health" in line:
        print(" ", line)
````

````python
"""
Section 8 -- Exercise both new endpoints through FastAPI's TestClient:
GET /emotion/labels and POST /emotion/predict.
"""
import sys as _sys
import base64
import numpy as np
import cv2

# Clean re-import so the patched main.py / new router are picked up fresh.
for mod_name in list(_sys.modules):
    if mod_name.startswith("backend") or mod_name.startswith("models.emotion_detection"):
        del _sys.modules[mod_name]
_sys.path.insert(0, str(project_root))

try:
    from backend.app.main import app
    from fastapi.testclient import TestClient

    client = TestClient(app)

    labels_resp = client.get("/emotion/labels")
    print(f"GET  /emotion/labels -> {labels_resp.status_code} ({len(labels_resp.json())} label(s))")
    assert labels_resp.status_code == 200
    assert len(labels_resp.json()) == 7

    rng = np.random.default_rng(4)
    crop = rng.integers(80, 180, size=(150, 130, 3), dtype=np.uint8)
    _, buffer = cv2.imencode(".jpg", crop)
    crop_b64 = base64.b64encode(buffer.tobytes()).decode("utf-8")

    predict_resp = client.post("/emotion/predict", json={
        "face_crop_b64": crop_b64, "student_id": "1BI23CS888",
    })
    print(f"POST /emotion/predict -> {predict_resp.status_code} {predict_resp.json()}")

    if predict_resp.status_code == 200:
        body = predict_resp.json()
        assert "emotion" in body and "confidence" in body and "processing_time" in body
        print("\nPASS: both emotion endpoints responded successfully.")
    elif predict_resp.status_code == 503:
        print("\n/emotion/predict returned 503 (DeepFace/TensorFlow not installed in this "
              "environment) -- endpoint wiring, routing, and request/response validation are "
              "still confirmed functional via /emotion/labels above.")
    else:
        raise AssertionError(f"Unexpected status code from /emotion/predict: {predict_resp.status_code}")

except ImportError as exc:
    print(f"FastAPI stack not fully available in this environment ({exc}). "
          "Router source code has still been written and is syntactically verified.")
````

# SECTION 12 — Updated Folder Structure

Everything below `models/emotion_detection/`, `configs/emotion_detection.yaml`,
`backend/app/routers/emotion.py`, and `tests/test_phase3_emotion.py` is new in Phase 3.
Every other Phase 1/2 path shown here already existed before this notebook ran.

````python
"""
Section 12 -- Display the updated project tree, highlighting what
Phase 3 added (same helper Phase 1/2 used).
"""
def print_tree(root, max_depth=2, prefix=""):
    """Pretty-print a directory tree up to max_depth levels (same helper Phase 1/2 used)."""
    if max_depth < 0:
        return
    entries = sorted([p for p in root.iterdir() if p.name != ".gitkeep"],
                      key=lambda p: (p.is_file(), p.name.lower()))
    for i, entry in enumerate(entries):
        connector = "\u2514\u2500\u2500 " if i == len(entries) - 1 else "\u251C\u2500\u2500 "
        print(f"{prefix}{connector}{entry.name}{'/' if entry.is_dir() else ''}")
        if entry.is_dir():
            extension = "    " if i == len(entries) - 1 else "\u2502   "
            print_tree(entry, max_depth=max_depth - 1, prefix=prefix + extension)


print(f"{project_root.name}/")
print_tree(project_root, max_depth=2)

print("\nNew in Phase 3:")
for new_path in [
    "models/emotion_detection/",
    "configs/emotion_detection.yaml",
    "backend/app/routers/emotion.py",
    "tests/test_phase3_emotion.py",
]:
    exists = (project_root / new_path).exists()
    print(f"  [{'OK' if exists else 'MISSING'}] {new_path}")
````
