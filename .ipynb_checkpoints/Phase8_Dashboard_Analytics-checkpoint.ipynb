# Predictive Multimodal Student Engagement and Cognitive Monitoring System
## Phase 8 -- Dashboard & Analytics

**Additive phase.** This notebook does **not** modify or regenerate anything
produced by Phases 1-7. It only *adds*:

- a new frontend single-page application (`frontend/`) that consumes
  existing Phase 2-7 APIs plus a small new aggregation layer,
- a new, purely additive backend module (`models/dashboard_analytics/`)
  and router (`backend/app/routers/dashboard.py`) that store and aggregate
  the JSON already produced by Phase 7's `/cognitive/predict`,
- one idempotent patch to `backend/app/main.py` that registers the new
  router and mounts the frontend as static files at `/dashboard-ui`.

No AI model, no prediction logic, and no existing router from Phases 1-7
is rewritten. Every write in this notebook is idempotent -- re-running the
notebook is always safe.

**Sections**

1. Project root detection & Phase 1-7 foundation verification
2. Phase 8 directory scaffolding
3. `dashboard_schemas.py` -- typed event/query contracts
4. `analytics_store.py` -- append-only event log + aggregation engine
5. `configs/dashboard.yaml` -- dashboard configuration
6. `backend/app/routers/dashboard.py` -- the dashboard API
7. Idempotent `main.py` patch (router + static mount)
8. Frontend: styles, utils, services, hooks, charts, components, layout, pages
9. Frontend validation (syntax, structure)
10. Backend AST & import validation
11. pytest suite
12. Phase 1-7 integrity re-verification
13. Project tree
14. Final summary


```python
"""
Environment setup, imports, and project-root detection.
"""
import ast
import hashlib
import importlib
import io
import json
import subprocess
import sys
import shutil
from pathlib import Path
from datetime import datetime, timezone

MARKER_FILES = [
    Path("backend") / "app" / "main.py",
    Path("backend") / "app" / "routers",
    Path("models"),
    Path("utils") / "common.py",
    Path("utils") / "logger.py",
]


def detect_project_root(start: Path = None) -> Path:
    """Walk upward from the notebook's working directory (or `start`)
    looking for the marker files that every completed phase leaves
    behind. Falls back to the current working directory with a warning
    if no marker set is found, so the notebook still runs standalone."""
    candidate = (start or Path.cwd()).resolve()
    for _ in range(6):
        if all((candidate / marker).exists() for marker in MARKER_FILES):
            return candidate
        if candidate.parent == candidate:
            break
        candidate = candidate.parent
    print("WARNING: could not locate a directory containing all Phase 1-7 "
          "marker files; falling back to the current working directory. "
          "If Phase 8 file paths look wrong, `cd` to the project root "
          "before launching Jupyter.")
    return (start or Path.cwd()).resolve()


PROJECT_ROOT = detect_project_root()
sys.path.insert(0, str(PROJECT_ROOT))
print(f"PROJECT_ROOT = {PROJECT_ROOT}")


def write_file_idempotent(path: Path, content: str) -> bool:
    """Write `content` to `path` only if the file does not already exist
    with byte-identical content. Returns True if a write happened."""
    path.parent.mkdir(parents=True, exist_ok=True)
    if path.exists() and path.read_text(encoding="utf-8") == content:
        return False
    path.write_text(content, encoding="utf-8")
    return True


def sha256_of(path: Path) -> str:
    return hashlib.sha256(path.read_bytes()).hexdigest() if path.exists() else ""


def print_section(title: str):
    print("\n" + "=" * 78)
    print(title)
    print("=" * 78)
```

## Section 1 -- Phase 1-7 Foundation Verification

Read-only. Records a SHA-256 baseline for every existing Phase 1-7 file so
Section 12 can prove nothing was modified.

```python
"""
Verify Phases 1-7 are present, and capture a checksum baseline so we can
prove later in this notebook that nothing from Phases 1-7 was modified.
"""
print_section("PHASE 1-7 FOUNDATION CHECK")

EXPECTED_ROUTERS = [
    "face_authentication.py",
    "emotion.py",
    "head_pose.py",
    "object_detection.py",
    "engagement_prediction.py",
    "cognitive_monitoring.py",
]

routers_dir = PROJECT_ROOT / "backend" / "app" / "routers"
found_routers = []
missing_routers = []
for name in EXPECTED_ROUTERS:
    path = routers_dir / name
    (found_routers if path.exists() else missing_routers).append(name)
    status = "PASS" if path.exists() else "WARN (not found)"
    print(f"  [{status:16}] backend/app/routers/{name}")

main_py = PROJECT_ROOT / "backend" / "app" / "main.py"
print(f"  [{'PASS' if main_py.exists() else 'FAIL':16}] backend/app/main.py")
assert main_py.exists(), "backend/app/main.py must exist -- is PROJECT_ROOT correct?"

models_dir = PROJECT_ROOT / "models"
print(f"  [{'PASS' if models_dir.is_dir() else 'FAIL':16}] models/")
assert models_dir.is_dir(), "models/ must exist -- is PROJECT_ROOT correct?"

if missing_routers:
    print(f"\nNOTE: {len(missing_routers)} expected router(s) not found by name "
          f"({missing_routers}). This is only a naming check -- Phase 8 does "
          f"not depend on it and will continue.")

# Baseline checksums for every existing Python file under backend/ and
# models/ (i.e. everything Phases 1-7 could have produced), *before*
# Phase 8 writes a single byte.
PHASE_1_7_BASELINE = {}
for base_dir in (PROJECT_ROOT / "backend", PROJECT_ROOT / "models"):
    if not base_dir.exists():
        continue
    for py_file in base_dir.rglob("*.py"):
        if "dashboard_analytics" in py_file.parts or py_file.name == "dashboard.py":
            continue  # these belong to Phase 8, not the Phase 1-7 baseline
        if py_file == main_py:
            continue  # main.py is intentionally appended to in Section 7; checked separately
        PHASE_1_7_BASELINE[str(py_file.relative_to(PROJECT_ROOT))] = sha256_of(py_file)

# main.py is the one existing file Phase 8 is allowed to touch, and only by
# appending -- so its pre-Phase-8 text is saved verbatim to prove later
# (Section 12) that every original line is still present, unmodified, at
# the start of the file.
MAIN_PY_ORIGINAL_TEXT = main_py.read_text(encoding="utf-8")

print(f"\nCaptured checksum baseline for {len(PHASE_1_7_BASELINE)} existing Phase 1-7 Python file(s).")
print("PASS  Phase 1-7 foundation check complete.")
```

## Section 4 -- `analytics_store.py`

Thread-safe, append-only JSON-Lines event log plus the aggregation engine
behind every stat card, chart, table, and report. Nothing here calls or
modifies a Phase 2-7 model; it only stores and aggregates the structured
results those phases already produce.

```python
"""
Write models/dashboard_analytics/analytics_store.py
"""
_target = PROJECT_ROOT / "models/dashboard_analytics/analytics_store.py"
_target.parent.mkdir(parents=True, exist_ok=True)
_content = '"""\nPhase 8 Dashboard & Analytics -- append-only event store and aggregation\nengine.\n\nEvery dashboard number (stat cards, timelines, student table, alerts,\nreports) is derived from the same append-only JSON-Lines event log at\n`data/dashboard/events.jsonl`. Nothing here calls or modifies any Phase\n2-7 model -- it only stores and aggregates the structured results those\nphases already produce (typically via Phase 7\'s `/cognitive/predict`\nresponse, forwarded into `POST /dashboard/events`).\n"""\nimport csv\nimport io\nimport json\nimport threading\nfrom collections import defaultdict\nfrom datetime import datetime, timedelta, timezone\nfrom pathlib import Path\nfrom typing import Any, Dict, List, Optional\n\nfrom models.dashboard_analytics.dashboard_schemas import DashboardEvent\nfrom utils.common import load_yaml_config\nfrom utils.logger import get_logger\n\nlogger = get_logger(__name__)\n\nDEFAULT_CONFIG_RELATIVE_PATH = Path("configs") / "dashboard.yaml"\nDEFAULT_EVENTS_RELATIVE_PATH = Path("data") / "dashboard" / "events.jsonl"\n\n_ALERT_SEVERITY = {\n    "PHONE_USAGE": "high",\n    "MULTIPLE_PERSON": "high",\n    "HIGH_DISTRACTION": "high",\n    "FATIGUE_DETECTED": "medium",\n    "LOOKING_AWAY": "medium",\n    "LOW_ATTENTION": "medium",\n    "LOW_CONFIDENCE": "low",\n    "UNKNOWN_FACE": "high",\n    "CONFUSION_DETECTED": "medium",\n    "DISENGAGED": "medium",\n}\n\n\nclass DashboardAnalyticsError(Exception):\n    """Base exception for all Phase 8 dashboard-analytics failures."""\n\n\ndef _parse_ts(value: str) -> datetime:\n    try:\n        dt = datetime.fromisoformat(value.replace("Z", "+00:00"))\n    except (ValueError, AttributeError):\n        return datetime.now(timezone.utc)\n    if dt.tzinfo is None:\n        dt = dt.replace(tzinfo=timezone.utc)\n    return dt\n\n\nclass DashboardAnalyticsStore:\n    """Thread-safe append-only event store plus read-side aggregation.\n\n    A future phase could swap the JSON-Lines file for a real database by\n    reimplementing just `append_event()` / `_read_events()` behind this\n    exact class interface -- every router endpoint keeps working\n    unmodified, mirroring the swappable-backend pattern Phase 6/7 already\n    use for their scoring engines.\n    """\n\n    def __init__(self, project_root: Path):\n        self.project_root = Path(project_root)\n        self._lock = threading.Lock()\n        self._config = self._load_config()\n        self.events_path = self.project_root / DEFAULT_EVENTS_RELATIVE_PATH\n        self.events_path.parent.mkdir(parents=True, exist_ok=True)\n        self.events_path.touch(exist_ok=True)\n\n    def _load_config(self) -> Dict[str, Any]:\n        config_path = self.project_root / DEFAULT_CONFIG_RELATIVE_PATH\n        try:\n            config = load_yaml_config(config_path)\n            if isinstance(config, dict) and config:\n                return config\n        except Exception as exc:  # noqa: BLE001 -- defensive fallback only\n            logger.warning(f"Falling back to default dashboard config: {exc}")\n        return {\n            "retention_days": 90,\n            "default_range": "today",\n            "timeline_bucket_minutes": 60,\n            "alert_thresholds": {"low_attention": 40.0, "high_distraction": 50.0, "fatigue": 60.0},\n            "theme": {"default_mode": "dark"},\n        }\n\n    def get_config(self) -> Dict[str, Any]:\n        return dict(self._config)\n\n    # -- write side ---------------------------------------------------\n    def append_event(self, event: DashboardEvent) -> None:\n        line = json.dumps(event.as_dict(), default=str)\n        with self._lock:\n            with self.events_path.open("a", encoding="utf-8") as fh:\n                fh.write(line + "\\n")\n        logger.info(f"Dashboard event recorded for student_id={event.student_id}")\n\n    # -- read side ------------------------------------------------------\n    def _read_events(\n        self,\n        since: Optional[datetime] = None,\n        student_id: Optional[str] = None,\n    ) -> List[Dict[str, Any]]:\n        events: List[Dict[str, Any]] = []\n        if not self.events_path.exists():\n            return events\n        with self._lock:\n            raw_lines = self.events_path.read_text(encoding="utf-8").splitlines()\n        for line in raw_lines:\n            line = line.strip()\n            if not line:\n                continue\n            try:\n                record = json.loads(line)\n            except json.JSONDecodeError:\n                continue\n            if student_id and record.get("student_id") != student_id:\n                continue\n            if since is not None and _parse_ts(record.get("timestamp", "")) < since:\n                continue\n            events.append(record)\n        return events\n\n    @staticmethod\n    def _range_start(range_: str) -> Optional[datetime]:\n        now = datetime.now(timezone.utc)\n        if range_ == "today":\n            return now.replace(hour=0, minute=0, second=0, microsecond=0)\n        if range_ == "week":\n            return now - timedelta(days=7)\n        if range_ == "month":\n            return now - timedelta(days=30)\n        return None  # "all"\n\n    # -- summary / stat cards --------------------------------------------\n    def compute_summary(self, range_: str = "today") -> Dict[str, Any]:\n        events = self._read_events(since=self._range_start(range_))\n        if not events:\n            return {\n                "today_attendance": 0,\n                "average_engagement": 0.0,\n                "average_attention": 0.0,\n                "average_confidence": 0.0,\n                "alerts_generated": 0,\n                "distraction_count": 0,\n                "phone_detections": 0,\n                "fatigue_detections": 0,\n                "event_count": 0,\n            }\n\n        distinct_students = {e["student_id"] for e in events if e.get("attendance")}\n        engagement_values = [e["engagement_score"] for e in events if e.get("engagement_score") is not None]\n        attention_values = [e["attention_score"] for e in events if e.get("attention_score") is not None]\n        confidence_values = [e["overall_confidence"] for e in events if e.get("overall_confidence") is not None]\n\n        alerts = [flag for e in events for flag in (e.get("risk_flags") or [])]\n        phone_detections = sum(1 for e in events if e.get("phone_detected"))\n        fatigue_detections = sum(1 for e in events if "FATIGUE_DETECTED" in (e.get("risk_flags") or []))\n        distraction_count = sum(1 for e in events if "HIGH_DISTRACTION" in (e.get("risk_flags") or []))\n\n        return {\n            "today_attendance": len(distinct_students),\n            "average_engagement": round(sum(engagement_values) / len(engagement_values), 2) if engagement_values else 0.0,\n            "average_attention": round(sum(attention_values) / len(attention_values), 2) if attention_values else 0.0,\n            "average_confidence": round(sum(confidence_values) / len(confidence_values), 4) if confidence_values else 0.0,\n            "alerts_generated": len(alerts),\n            "distraction_count": distraction_count,\n            "phone_detections": phone_detections,\n            "fatigue_detections": fatigue_detections,\n            "event_count": len(events),\n        }\n\n    # -- timelines --------------------------------------------------------\n    def compute_timeline(self, metric: str, range_: str = "today") -> List[Dict[str, Any]]:\n        valid_metrics = (\n            "emotion", "engagement", "cognitive", "attention", "fatigue",\n            "phone_detection", "multiple_people", "attendance", "risk_flags",\n        )\n        if metric not in valid_metrics:\n            raise DashboardAnalyticsError(f"Unsupported timeline metric: {metric!r}")\n\n        events = sorted(\n            self._read_events(since=self._range_start(range_)),\n            key=lambda e: e.get("timestamp", ""),\n        )\n        bucket_minutes = int(self._config.get("timeline_bucket_minutes", 60))\n        buckets: Dict[str, List[float]] = defaultdict(list)\n        bucket_meta: Dict[str, Dict[str, int]] = defaultdict(lambda: defaultdict(int))\n\n        for record in events:\n            ts = _parse_ts(record.get("timestamp", ""))\n            epoch_minutes = int(ts.timestamp() // 60)\n            bucket_key_minutes = epoch_minutes - (epoch_minutes % bucket_minutes)\n            bucket_dt = datetime.fromtimestamp(bucket_key_minutes * 60, tz=timezone.utc)\n            bucket_key = bucket_dt.isoformat()\n\n            if metric == "emotion":\n                bucket_meta[bucket_key][str(record.get("emotion") or "unknown")] += 1\n            elif metric == "engagement":\n                buckets[bucket_key].append(float(record.get("engagement_score", 0.0)))\n            elif metric == "cognitive":\n                bucket_meta[bucket_key][str(record.get("cognitive_state") or "unknown")] += 1\n            elif metric == "attention":\n                buckets[bucket_key].append(float(record.get("attention_score", 0.0)))\n            elif metric == "fatigue":\n                buckets[bucket_key].append(float(record.get("fatigue_score", 0.0)))\n            elif metric == "phone_detection":\n                buckets[bucket_key].append(1.0 if record.get("phone_detected") else 0.0)\n            elif metric == "multiple_people":\n                buckets[bucket_key].append(1.0 if record.get("multiple_person") else 0.0)\n            elif metric == "attendance":\n                buckets[bucket_key].append(1.0 if record.get("attendance") else 0.0)\n            elif metric == "risk_flags":\n                for flag in record.get("risk_flags") or []:\n                    bucket_meta[bucket_key][flag] += 1\n            else:\n                raise DashboardAnalyticsError(f"Unsupported timeline metric: {metric!r}")\n\n        if metric in ("emotion", "cognitive", "risk_flags"):\n            return [\n                {"timestamp": key, "counts": dict(counts)}\n                for key, counts in sorted(bucket_meta.items())\n            ]\n\n        return [\n            {"timestamp": key, "value": round(sum(values) / len(values), 3) if values else 0.0}\n            for key, values in sorted(buckets.items())\n        ]\n\n    # -- students table ---------------------------------------------------\n    def compute_student_table(self) -> List[Dict[str, Any]]:\n        events = self._read_events()\n        latest_by_student: Dict[str, Dict[str, Any]] = {}\n        for record in events:\n            student_id = record.get("student_id")\n            if not student_id:\n                continue\n            existing = latest_by_student.get(student_id)\n            if existing is None or record.get("timestamp", "") >= existing.get("timestamp", ""):\n                latest_by_student[student_id] = record\n\n        rows = []\n        now = datetime.now(timezone.utc)\n        for student_id, record in latest_by_student.items():\n            last_seen = _parse_ts(record.get("timestamp", ""))\n            status = "active" if (now - last_seen) <= timedelta(minutes=10) else "offline"\n            rows.append({\n                "student_id": student_id,\n                "attendance": bool(record.get("attendance", False)),\n                "last_emotion": record.get("emotion"),\n                "engagement_score": record.get("engagement_score", 0.0),\n                "engagement_level": record.get("engagement_level"),\n                "cognitive_state": record.get("cognitive_state"),\n                "last_seen": record.get("timestamp"),\n                "status": status,\n            })\n        return sorted(rows, key=lambda r: r["student_id"])\n\n    # -- alerts -------------------------------------------------------------\n    def compute_alerts(self, range_: str = "today", flag: Optional[str] = None) -> List[Dict[str, Any]]:\n        events = self._read_events(since=self._range_start(range_))\n        alerts: List[Dict[str, Any]] = []\n        for record in events:\n            for risk_flag in record.get("risk_flags") or []:\n                if flag and risk_flag != flag:\n                    continue\n                alerts.append({\n                    "student_id": record.get("student_id"),\n                    "timestamp": record.get("timestamp"),\n                    "flag": risk_flag,\n                    "severity": _ALERT_SEVERITY.get(risk_flag, "low"),\n                    "cognitive_state": record.get("cognitive_state"),\n                })\n        return sorted(alerts, key=lambda a: a["timestamp"], reverse=True)\n\n    # -- reports --------------------------------------------------------------\n    def compute_report(self, period: str, student_id: Optional[str] = None) -> Dict[str, Any]:\n        range_map = {"daily": "today", "weekly": "week", "monthly": "month", "student": "all", "class": "all"}\n        range_ = range_map.get(period, "all")\n        events = self._read_events(since=self._range_start(range_), student_id=student_id)\n\n        durations: Dict[str, List[datetime]] = defaultdict(list)\n        for record in events:\n            durations[record["student_id"]].append(_parse_ts(record.get("timestamp", "")))\n\n        session_durations = {\n            sid: round((max(ts_list) - min(ts_list)).total_seconds() / 60.0, 1)\n            for sid, ts_list in durations.items()\n        }\n\n        return {\n            "period": period,\n            "student_id": student_id,\n            "summary": self.compute_summary(range_=range_),\n            "student_count": len({e["student_id"] for e in events}),\n            "session_duration_minutes": session_durations,\n            "event_count": len(events),\n        }\n\n    def export_csv(self, range_: str = "all") -> str:\n        events = self._read_events(since=self._range_start(range_))\n        buffer = io.StringIO()\n        fieldnames = [\n            "timestamp", "student_id", "attendance", "emotion", "cognitive_state",\n            "attention_score", "engagement_score", "engagement_level",\n            "phone_detected", "multiple_person", "overall_confidence", "risk_flags",\n        ]\n        writer = csv.DictWriter(buffer, fieldnames=fieldnames, extrasaction="ignore")\n        writer.writeheader()\n        for record in events:\n            row = dict(record)\n            row["risk_flags"] = ";".join(record.get("risk_flags") or [])\n            writer.writerow(row)\n        return buffer.getvalue()\n\n\n_STORE_SINGLETON: Optional[DashboardAnalyticsStore] = None\n_SINGLETON_LOCK = threading.Lock()\n\n\ndef get_analytics_store(project_root: Optional[Path] = None) -> DashboardAnalyticsStore:\n    """Return the process-wide `DashboardAnalyticsStore` singleton, creating\n    it on first call (mirrors Phase 6/7\'s `get_*_predictor()` /\n    `get_cognitive_monitor()` singleton pattern)."""\n    global _STORE_SINGLETON\n    with _SINGLETON_LOCK:\n        if _STORE_SINGLETON is None:\n            if project_root is None:\n                # backend/app/routers/dashboard.py -> parents[2] == project root\n                project_root = Path(__file__).resolve().parents[2]\n            _STORE_SINGLETON = DashboardAnalyticsStore(project_root)\n        return _STORE_SINGLETON\n\n\ndef reset_analytics_store() -> None:\n    """Test-only helper: forces the next `get_analytics_store()` call to\n    build a fresh singleton (used by the pytest suite between cases)."""\n    global _STORE_SINGLETON\n    with _SINGLETON_LOCK:\n        _STORE_SINGLETON = None\n'
write_file_idempotent(_target, _content)
print(f"PASS  wrote {_target.relative_to(PROJECT_ROOT)}  ({len(_content)} chars)")
```

(assembles the shell) -> `pages/` (Dashboard Home, Analytics, Reports,
Students, Alerts, Settings, and the hash router) -> `main.js` + `index.html`.

## Section 9 -- Frontend Validation

Every JS file is syntax-checked with Node.js (`node --check`) when Node is
available, falling back to a brace/paren-balance check otherwise. Every
CSS file is checked for balanced braces. `index.html` is checked for the
mount point and module script tag it needs.

```python
"""
Validate every generated frontend file.
"""
print_section("FRONTEND VALIDATION")

import subprocess as _subprocess

_frontend_dir = PROJECT_ROOT / "frontend"
_js_files = sorted(_frontend_dir.rglob("*.js"))
_css_files = sorted(_frontend_dir.rglob("*.css"))

_node_available = shutil.which("node") is not None
print(f"Node.js available for syntax checking: {_node_available}\n")

_js_failures = []
for js_file in _js_files:
    rel = js_file.relative_to(PROJECT_ROOT)
    if _node_available:
        result = _subprocess.run(["node", "--check", str(js_file)], capture_output=True, text=True)
        ok = result.returncode == 0
        detail = "" if ok else result.stderr.strip().splitlines()[-1]
    else:
        text = js_file.read_text(encoding="utf-8")
        ok = text.count("{") == text.count("}") and text.count("(") == text.count(")")
        detail = "" if ok else "unbalanced braces/parens"
    print(f"  [{'PASS' if ok else 'FAIL':6}] {rel}" + (f"  -- {detail}" if detail else ""))
    if not ok:
        _js_failures.append(str(rel))

_css_failures = []
for css_file in _css_files:
    rel = css_file.relative_to(PROJECT_ROOT)
    text = css_file.read_text(encoding="utf-8")
    ok = text.count("{") == text.count("}")
    print(f"  [{'PASS' if ok else 'FAIL':6}] {rel}")
    if not ok:
        _css_failures.append(str(rel))

_index_html = (_frontend_dir / "index.html").read_text(encoding="utf-8")
_html_ok = ('id="app"' in _index_html) and ("main.js" in _index_html) and ('type="module"' in _index_html)
print(f"  [{'PASS' if _html_ok else 'FAIL':6}] frontend/index.html has #app mount point + module entry script")

assert not _js_failures, f"JS validation failed for: {_js_failures}"
assert not _css_failures, f"CSS validation failed for: {_css_failures}"
assert _html_ok, "index.html is missing required structure"
print(f"\nPASS  all {len(_js_files)} JS file(s), {len(_css_files)} CSS file(s), and index.html are valid.")
```

## Section 13 -- Project Tree

```python
"""
Print the Phase 8 additions to the project tree.
"""
print_section("PROJECT TREE -- PHASE 8 ADDITIONS")


def print_tree(base: Path, prefix: str = ""):
    entries = sorted([p for p in base.iterdir() if p.name not in
                       {"__pycache__", ".git", "node_modules", ".pytest_cache"}],
                      key=lambda p: (p.is_file(), p.name.lower()))
    for i, entry in enumerate(entries):
        connector = "\u2514\u2500\u2500 " if i == len(entries) - 1 else "\u251c\u2500\u2500 "
        print(prefix + connector + entry.name + ("/" if entry.is_dir() else ""))
        if entry.is_dir():
            extension = "    " if i == len(entries) - 1 else "\u2502   "
            print_tree(entry, prefix + extension)


print(f"{PROJECT_ROOT.name}/")
print("frontend/")
print_tree(PROJECT_ROOT / "frontend", "")
print()
print("models/dashboard_analytics/")
print_tree(PROJECT_ROOT / "models" / "dashboard_analytics", "")
print()
print("New/updated backend + config files:")
for rel in [
    "backend/app/routers/dashboard.py",
    "backend/app/main.py  (patched, additive only)",
    "configs/dashboard.yaml",
    "tests/test_phase8_dashboard_analytics.py",
    "data/dashboard/events.jsonl  (created on first recorded event)",
]:
    print(f"  {rel}")
```
