```python
from pathlib import Path

# Root folder
root = Path("backend")

# Folders to create
folders = [
    root,
    root / "routes",
    root / "services",
    root / "database",
    root / "models",
    root / "utils"
]

# Create folders
for folder in folders:
    folder.mkdir(parents=True, exist_ok=True)

# Files to create
files = [
    root / "app.py",
    root / "config.py",
    root / "requirements.txt",

    root / "routes" / "__init__.py",
    root / "routes" / "auth.py",
    root / "routes" / "teacher.py",
    root / "routes" / "student.py",
    root / "routes" / "monitoring.py",

    root / "services" / "__init__.py",
    root / "services" / "ai_service.py",
    root / "services" / "report_service.py",

    root / "database" / "__init__.py",
    root / "database" / "database.py",

    root / "models" / "__init__.py",
    root / "models" / "student.py",
    root / "models" / "teacher.py",
    root / "models" / "session.py",
    root / "models" / "engagement.py",

    root / "utils" / "helpers.py"
]

# Create empty files
for file in files:
    file.touch(exist_ok=True)

print("=" * 60)
print("✅ Backend project structure created successfully!")
print("=" * 60)

# Display folder structure
def print_tree(path, indent=""):
    items = sorted(path.iterdir(), key=lambda x: (x.is_file(), x.name))
    for item in items:
        print(indent + ("📁 " if item.is_dir() else "📄 ") + item.name)
        if item.is_dir():
            print_tree(item, indent + "    ")

print_tree(root)
```

```python
from pathlib import Path

code = '''from sqlalchemy.orm import Session

from models.student import Student


class StudentRepository:

    @staticmethod
    def create_student(db: Session, student: Student):
        db.add(student)
        db.commit()
        db.refresh(student)
        return student


    @staticmethod
    def get_student_by_email(db: Session, email: str):
        return (
            db.query(Student)
            .filter(Student.email == email)
            .first()
        )


    @staticmethod
    def get_student_by_usn(db: Session, usn: str):
        return (
            db.query(Student)
            .filter(Student.usn == usn)
            .first()
        )


    @staticmethod
    def get_student_by_id(db: Session, student_id: int):
        return (
            db.query(Student)
            .filter(Student.student_id == student_id)
            .first()
        )
'''

Path("backend/repositories/student_repository.py").write_text(
    code,
    encoding="utf-8"
)

print("✅ student_repository.py created successfully.")
```

```python
from pathlib import Path

code = '''import bcrypt


class SecurityService:

    @staticmethod
    def hash_password(password: str) -> str:
        return bcrypt.hashpw(
            password.encode("utf-8"),
            bcrypt.gensalt()
        ).decode("utf-8")


    @staticmethod
    def verify_password(
        plain_password: str,
        hashed_password: str
    ) -> bool:

        return bcrypt.checkpw(
            plain_password.encode("utf-8"),
            hashed_password.encode("utf-8")
        )
'''

Path("backend/services/security_service.py").write_text(
    code,
    encoding="utf-8"
)

print("✅ security_service.py created successfully.")
```

```python
from pathlib import Path

code = '''import cv2
import numpy as np
import face_recognition


class FaceService:

    @staticmethod
    def generate_face_embedding(image):

        rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        locations = face_recognition.face_locations(rgb)

        if len(locations) == 0:
            raise Exception("No face detected.")

        if len(locations) > 1:
            raise Exception("Multiple faces detected.")

        encoding = face_recognition.face_encodings(
            rgb,
            locations
        )[0]

        return encoding.tolist()

    @staticmethod
    def compare_faces(stored_embedding, current_embedding):

        result = face_recognition.compare_faces(
            [np.array(stored_embedding)],
            np.array(current_embedding),
            tolerance=0.45
        )

        return result[0]
'''

Path("backend/services/face_service.py").write_text(code)

print("✅ face_service.py created successfully.")
```
